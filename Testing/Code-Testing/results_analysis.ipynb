```python
combined_errors_subfull_epochs = pd.concat([sub_subepoch_erros_df, results_errors_sub], ignore_index=True)
combined_errors_subfull_epochs.head()
```

```python
from sklearn.neighbors import KernelDensity

import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.gridspec as grid_spec
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


nframes = [200, 400, 800, 1400]
colors = ['#0000ff', '#660099', '#990066', '#ff0000']

gs = grid_spec.GridSpec(4,1)
fig = plt.figure(figsize=(16,9))

i = 0

ax_objs = []
for frame_count in nframes:
    
    x = np.array(combined_errors_subfull_epochs[combined_errors_subfull_epochs.numframes == frame_count].distance_px)
    x_d = np.linspace(0, 15, 400)

    kde = KernelDensity(bandwidth=0.03, kernel='gaussian')
    kde.fit(x[:, None])

    logprob = kde.score_samples(x_d[:, None])

    # creating new axes object
    ax_objs.append(fig.add_subplot(gs[i:i+1, 0:]))

    # plotting the distribution
    ax_objs[-1].plot(x_d, np.exp(logprob), color="#f0f0f0",lw=1)
    ax_objs[-1].fill_between(x_d, np.exp(logprob), alpha=1, color = colors[i])


    # setting uniform x and y lims
    ax_objs[-1].set_xlim(0,15)
    ax_objs[-1].set_ylim(0,2.5)

    # make background transparent
    rect = ax_objs[-1].patch
    rect.set_alpha(0)

    # remove borders, axis ticks, and labels
    ax_objs[-1].set_yticklabels([])

    if i == len(nframes)-1:
        ax_objs[-1].set_xlabel("Test Score", fontsize=16,fontweight="bold")
    else:
        ax_objs[-1].set_xticklabels([])

    spines = ["top","right","left","bottom"]
    for s in spines:
        ax_objs[-1].spines[s].set_visible(False)

    ax_objs[-1].text(-0.02, 0, frame_count, fontweight="bold", fontsize=14, ha="right")


    i += 1

gs.update(hspace=-0.7)

fig.text(0.07,0.85,"Distribution of Pixel Errors",fontsize=20)

plt.tight_layout()
plt.show()

```

```python
# distance KDE
nframes = [200, 400, 800, 1400]
colors = ['#0000ff', '#660099', '#990066', '#ff0000']

x_err = results_errors.loc[
    results_errors.num_frames == frame_count,
    "distance_px"
].dropna()

# likelihood KDE
x_like = results_errors.loc[
    results_errors.num_frames == frame_count,
    "likelihood"
].dropna()

# kde_err = KernelDensity(bandwidth=0.03, kernel='gaussian')
# kde_err.fit(x_err.to_numpy()[:, None])

# kde_like = KernelDensity(bandwidth=0.03, kernel='gaussian')
# kde_like.fit(x_like.to_numpy()[:, None])

# err_grid = np.linspace(0, 15, 400)
# like_grid = np.linspace(0, 1, 400)

# err_density = np.exp(
#     kde_err.score_samples(err_grid[:, None])
# )

# like_density = np.exp(
#     kde_like.score_samples(like_grid[:, None])
# )
i = 0
gs = grid_spec.GridSpec(4,2)
fig = plt.figure(figsize=(16,9))

for frame_count in nframes:
    
    x_err = results_errors.loc[
        results_errors.num_frames == frame_count,
        "distance_px"
    ].dropna()

    # likelihood KDE
    x_like = results_errors.loc[
        results_errors.num_frames == frame_count,
        "likelihood"
    ].dropna()
    kde_err = KernelDensity(bandwidth=0.03, kernel='gaussian')
    kde_err.fit(x_err.to_numpy()[:, None])

    kde_like = KernelDensity(bandwidth=0.03, kernel='gaussian')
    kde_like.fit(x_like.to_numpy()[:, None])

    err_grid = np.linspace(0, 15, 400)
    like_grid = np.linspace(0, 1, 400)

    err_density = np.exp(
        kde_err.score_samples(err_grid[:, None])
    )

    like_density = np.exp(kde_like.score_samples(like_grid[:, None]))
    
    # LEFT side = error
    ax_objs[-1].fill_between(
        -err_grid,
        err_density,
        color=colors[i],
        alpha=0.8
    )

    # RIGHT side = likelihood
    ax_objs[-1].fill_between(
        like_grid,
        like_density,
        color=colors[i],
        alpha=0.4
    )

    ax_objs[-1].axvline(0, color="black", lw=1)
    ax_objs[-1].set_xlim(-15, 1)
    ax_objs[-1].set_xticks([-15, -10, -5, 0, 0.5, 1])

    ax_objs[-1].set_xticklabels([
        "15",
        "10",
        "5",
        "0",
        "0.5",
        "1.0"
    ])
    i += 1
gs.update(hspace=-0.7)

fig.text(0.07,0.85,"Distribution of Pixel Errors",fontsize=20)

plt.tight_layout()
plt.show()
```
