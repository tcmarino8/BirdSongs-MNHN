```python
# add leading 00 to jpgs in folder
def add_leading_zeros_to_jpgs(input_folder: pathlib.Path, prefix: str, total_digits: int):
    for img_path in input_folder.glob("*.jpg"):
        stem = img_path.stem
        match = re.match(rf"{re.escape(prefix)}(\d+)", stem)
        if match:
            number_str = match.group(1)
            new_number_str = number_str.zfill(total_digits)
            new_stem = f"{prefix}{new_number_str}"
            new_img_path = img_path.with_name(new_stem + img_path.suffix)
            img_path.rename(new_img_path)

# remove second Img from cam2_ImgImg00_UND.0001.jpg -> cam2_Img00_UND.0001.jpg
def remove_redundant_img(input_folder: pathlib.Path):
    for img_path in input_folder.glob("*.jpg"):
        stem = img_path.stem
        new_stem = stem.replace("ImgImg", "Img")
        new_img_path = img_path.with_name(new_stem + img_path.suffix)
        img_path.rename(new_img_path)

# remove 00_UND. from image stack
def remove_00_UND_from_jpgs(input_folder: pathlib.Path):
    for img_path in input_folder.glob("*.jpg"):
        stem = img_path.stem
        new_stem = stem.replace("00_UND.", "")
        new_img_path = img_path.with_name(new_stem + img_path.suffix)
        img_path.rename(new_img_path)

def add_redundant_img(input_folder: pathlib.Path):
    for img_path in input_folder.glob("*.jpg"):
        stem = img_path.stem
        new_stem = stem.replace("Img", "ImgImg")
        new_img_path = img_path.with_name(new_stem + img_path.suffix)
        img_path.rename(new_img_path)



# Data writen with repeated T15_T15 in index, e.g. labeled-data/DavidBowie_T15_T15_n200/TrainingData_T15_cam1_0088.png
def _normalize_labeled_index(index_value: str, bird: str, trial: str, nframes: str) -> str:
    s = str(index_value).replace("\\", "/")
    s = s.replace("Trep_UND", "")

    # If index has duplicated trial tags in dataset folder, collapse them first:
    # labeled-data/DavidBowie_T15_T15_n200/... -> labeled-data/DavidBowie_T15_n200/...
    dup_match = re.search(
        rf"labeled-data/{re.escape(bird)}_T(?P<t1>\d+)_T(?P<t2>\d+)_n(?P<n>\d+)/",
        s,
    )
    if dup_match and dup_match.group("t1") == dup_match.group("t2"):
        detected_trial = dup_match.group("t1")
        detected_nframes = dup_match.group("n")
        s = re.sub(
            r"labeled-data/[^/]+/",
            f"labeled-data/{bird}_T{detected_trial}_n{detected_nframes}/",
            s,
        )
    else:
        # fallback to expected trial-specific folder from project metadata
        s = re.sub(
            r"labeled-data/[^/]+/",
            f"labeled-data/{bird}_T{trial}_n{nframes}/",
            s,
        )

    # remove embedded trial token from frame filename, e.g.
    # TrainingData_T15_cam1_0088.png -> TrainingData_cam1_0088.png
    s = re.sub(r"TrainingData_T\d+_(cam[12]_\d+\.(?:png|jpg|jpeg))$", r"TrainingData_\1", s)
    s = re.sub(r"//+", "/", s)
    return s

def _rewrite_collected_data_index(
    file_path: pathlib.Path,
    bird: str,
    trial: str,
    nframes: str,
    dry_run: bool = True,
    preview_limit: int = 3,
 ) -> tuple[int, int]:
    ext = file_path.suffix.lower()
    changed = 0

    if ext == ".csv":
        df = pd.read_csv(file_path, header=[0, 1, 2], index_col=0)
        old_index = df.index.astype(str).tolist()
        new_index = [_normalize_labeled_index(x, bird, trial, nframes) for x in old_index]
        changed = sum(o != n for o, n in zip(old_index, new_index))

        if changed and preview_limit > 0:
            shown = 0
            for old, new in zip(old_index, new_index):
                if old != new:
                    print(f"    preview: {old} -> {new}")
                    shown += 1
                    if shown >= preview_limit:
                        break

        if changed and not dry_run:
            df.index = new_index
            df.to_csv(file_path)
        return len(df.index), changed

    if ext == ".h5":
        df = pd.read_hdf(file_path)
        old_index = df.index.astype(str).tolist()
        new_index = [_normalize_labeled_index(x, bird, trial, nframes) for x in old_index]
        changed = sum(o != n for o, n in zip(old_index, new_index))

        if changed and preview_limit > 0:
            shown = 0
            for old, new in zip(old_index, new_index):
                if old != new:
                    print(f"    preview: {old} -> {new}")
                    shown += 1
                    if shown >= preview_limit:
                        break

        if changed and not dry_run:
            df.index = new_index
            df.to_hdf(file_path, key="df_with_missing", mode="w")
        return len(df.index), changed

    return 0, 0

def remove_Trep_from_data(
    input_folder: pathlib.Path,
    dry_run: bool = True,
    preview_limit: int = 3,
 ):
    # input_folder can be either bird root (.../DavidBowie) or Model519 root.
    # dry_run=True previews planned index updates without writing any files.
    model_root = input_folder / "Model519" if (input_folder / "Model519").exists() else input_folder
    if not model_root.exists():
        raise FileNotFoundError(f"Path not found: {model_root}")

    mode = "DRY-RUN" if dry_run else "APPLY"
    print(f"[{mode}] scanning {model_root}")

    project_dirs = [p for p in model_root.iterdir() if p.is_dir() and re.match(r".+_n\d+_T\d+$", p.name)]
    total_files = 0
    total_changed_rows = 0

    for project_dir in sorted(project_dirs):
        m = re.match(r"(?P<bird>.+)_n(?P<nframes>\d+)_T(?P<trial>\d+)$", project_dir.name)
        if not m:
            continue

        bird = m.group("bird")
        nframes = m.group("nframes")
        trial = m.group("trial")

        run_dirs = [p.parent for p in project_dir.glob("*/config.yaml")]
        if not run_dirs:
            continue

        run_dir = sorted(run_dirs)[-1]
        labeled_data_root = run_dir / "labeled-data"
        if not labeled_data_root.exists():
            continue

        csv_files = sorted(labeled_data_root.glob("**/CollectedData_*.csv"))
        h5_files = sorted(labeled_data_root.glob("**/CollectedData_*.h5"))

        for file_path in csv_files + h5_files:
            rows, changed = _rewrite_collected_data_index(
                file_path=file_path,
                bird=bird,
                trial=trial,
                nframes=nframes,
                dry_run=dry_run,
                preview_limit=preview_limit,
            )
            total_files += 1
            total_changed_rows += changed
            action = "would change" if dry_run else "changed"
            print(f"[{project_dir.name}] {file_path.name}: {action} {changed}/{rows} index rows")

    verb = "would update" if dry_run else "updated"
    print(f"Done. Processed {total_files} files, {verb} {total_changed_rows} index rows.")
```

```python
# Do the same as above but for images in stacks/labeled-data, e.g.
# Cam1_Img00_Trep_UND.0001.jpg -> Cam1_Img00.0001.jpg
# TrainingData_T32_cam1_0001.png -> TrainingData_cam1_0001.png
def remove_Trep_from_jpgs(
    input_folder: pathlib.Path,
    dry_run: bool = True,
    preview_limit: int = 10,
    recursive: bool = False,
 ) -> tuple[int, int]:
    if not input_folder.exists():
        raise FileNotFoundError(f"Path not found: {input_folder}")

    mode = "DRY-RUN" if dry_run else "APPLY"
    print(f"[{mode}] scanning {input_folder}")

    pattern_iter = input_folder.rglob("*") if recursive else input_folder.glob("*")
    image_paths = [
        p for p in pattern_iter
        if p.is_file() and p.suffix.lower() in {".jpg", ".jpeg", ".png"}
    ]

    total = len(image_paths)
    changed = 0
    shown = 0

    for img_path in image_paths:
        stem = img_path.stem

        # 1) remove Trep token
        new_stem = stem.replace("Trep_UND", "")

        # 2) collapse TrainingData_TXX_camY_#### -> TrainingData_camY_####
        new_stem = re.sub(r"^TrainingData_T\d+_(cam[12]_\d+)$", r"TrainingData_\1", new_stem)

        if new_stem == stem:
            continue

        new_img_path = img_path.with_name(new_stem + img_path.suffix)
        changed += 1

        if shown < preview_limit:
            action = "would rename" if dry_run else "renamed"
            print(f"  {action}: {img_path.name} -> {new_img_path.name}")
            shown += 1

        if not dry_run:
            img_path.rename(new_img_path)

    verb = "would rename" if dry_run else "renamed"
    print(f"Done. Scanned {total} images, {verb} {changed} files.")
    return total, changed

# Example on the path you shared (folder-level):
# remove_Trep_from_jpgs(
#     input_folder=pathlib.Path(r"C:\Users\Salle-Cineradio\Documents\MachineLearning\BirdSongs-MNHN\Testing\DeepLabCut\Endive\Model519\Endive_n400_T32\Canari-Tyler-2026-05-19\labeled-data\Endive_T32_n400"),
#     dry_run=True,
#     recursive=False,
# )
```

```python
# Run remove_Trep_from_jpgs for all birds from the bird-root folder
bird_root = pathlib.Path(
    r"C:\Users\Salle-Cineradio\Documents\MachineLearning\BirdSongs-MNHN\Testing\DeepLabCut"
)

dry_run = True
preview_limit = 5

total_scanned = 0
total_changed = 0
datasets_processed = 0

for bird_dir in sorted(bird_root.iterdir()):
    if not bird_dir.is_dir():
        continue

    # Search each bird folder for DLC labeled-data dataset folders.
    # Example target: .../labeled-data/Endive_T32_n400
    dataset_dirs = [p for p in bird_dir.glob("**/labeled-data/*") if p.is_dir()]
    if not dataset_dirs:
        continue

    print(f"\n=== Bird: {bird_dir.name} ({len(dataset_dirs)} dataset folders) ===")
    for dataset_dir in sorted(dataset_dirs):
        scanned, changed = remove_Trep_from_jpgs(
            input_folder=dataset_dir,
            dry_run=dry_run,
            preview_limit=preview_limit,
            recursive=False,
        )
        total_scanned += scanned
        total_changed += changed
        datasets_processed += 1

verb = "would rename" if dry_run else "renamed"
print(f"\nDone. Processed {datasets_processed} dataset folders.")
print(f"Total scanned images: {total_scanned}")
print(f"Total files that {verb}: {total_changed}")
```
