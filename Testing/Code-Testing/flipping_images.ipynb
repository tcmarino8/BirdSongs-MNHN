```python
from pathlib import Path
from PIL import Image

# Choose flip type: "horizontal" or "vertical"
flip_method = "horizontal"

base_dir = Path(r"C:\Users\Salle-Cineradio\Documents\MachineLearning\BirdSongs-MNHN\Testing\ProcessingData\Pig\Trial6")

# Process both folders
for cam in ["Cam1", "Cam2"]:
    folder = base_dir / cam

    for img_path in folder.glob("*"):
        if img_path.suffix.lower() not in {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}:
            continue

        with Image.open(img_path) as img:
            if flip_method == "horizontal":
                flipped = img.transpose(Image.FLIP_LEFT_RIGHT)
            elif flip_method == "vertical":
                flipped = img.transpose(Image.FLIP_TOP_BOTTOM)
            else:
                raise ValueError("flip_method must be 'horizontal' or 'vertical'")

            # Overwrite the original image
            flipped.save(img_path)


print("Done!")
```
